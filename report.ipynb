Now it's time to present my measure of similarity of works, on a scale 0-100 how similar are two works.
Imagine that for each work you have a list of all the words in it and how often they appeared. Then take the top $n$ of those and compute how often they occur (let $w_i$ be the number of appearances of $i$th most popular word), i.e. for each word divide its cardinality by the number of all words in a work (let's call it $N$). That will create a sequence of numbers
$$ u_1 = \frac{w_1}{N}, u_2 = \frac{w_2}{N},..., u_n = \frac{w_n}{N}, $$
where $u_1 \geqslant u_2 \geqslant ... \geqslant u_n$ and $\sum_{i=1}^n u_i \leqslant 1$.
Such sequence is a probability distribution cutted at first n values.
To compare 2 works we need to have such sequence for both works, let's call them
$u^{(1)} = (u^{(1)}_i)^n_{i=1}$ and $u^{(2)} = (u^{(2)}_i)^n_{i=1}$.
Our measure will be a convex combination of two different measures of similarity of those two sequences. The first one is $L^1$ norm of their difference:
$$ ||u^{(1)} - u^{(2)}||_{L^1} = \sum_{i=1}^n |u^{(1)}_i - u^{(2)}_i| $$
Notice that $|u^{(1)}_i - u^{(2)}_i| \leqslant |u^{(1)}_i| + |u^{(2)}_i| = u^{(1)}_i + u^{(2)}_i$, so
$$ ||u^{(1)} - u^{(2)}||_{L^1} = \sum_{i=1}^n |u^{(1)}_i - u^{(2)}_i| \leqslant \sum_{i=1}^n u^{(1)}_i + u^{(2)}_i \leqslant 1+1 = 2$$,
therefore we will use
$\frac{1}{2} ||u^{(1)} - u^{(2)}||_{L^1}$. Obviously
$1 - \frac{1}{2} ||u^{(1)} - u^{(2)}||_{L^1} = 1$ if and only if $u^{(1)} = u^{(2)}$.
The next metric is called **Jaccard index**. It measures how two sets are similar in shape. In its most basic form for two sets $A, B$ it's defined as
$$ J(A,B) = \frac{|A \cap B|}{|A \cup B|}. $$
Notice that $0 \leqslant J(A,B) \leqslant 1$ and $J(A,B) = 1$ if and only if $A=B$. We will use what its called **weighted Jaccard index** and its defined as
$$ J_W(u^{(1)} , u^{(2)}) = \frac{\sum_{i=1}^n \min(u^{(1)}_i , u^{(2)}_i)}{\sum_{i=1}^n \max(u^{(1)}_i , u^{(2)}_i)}. $$
We can imagine this as sum of Jaccard indexes between two rectangles: one has $1 \times u^{(1)}_i$ dimensions and the other are $1 \times u^{(2)}_i.$
In total, our similarity mesaure is
$$ m(u^{(1)} , u^{(2)}) = 100\left(  \alpha  J_W(u^{(1)} , u^{(2)}) + \left( 1 - \alpha \right) \left( 1 - \frac{1}{2} ||u^{(1)} - u^{(2)}||_{L^1} \right) \right). $$ Summing up all of our observations, we can see that
$0 \leqslant m(u^{(1)} , u^{(2)}) \leqslant 100$ and if and $m(u^{(1)} , u^{(2)}) = 100$ only if $u^{(1)} = u^{(2)}$.
We will use $\alpha = 0.5$ as deafult. Let us now compute how similar each work is to the other.
